# Homework 9 – Face Detection with dlib
In this notebook we build a face detector using dlib and test it on two images:
1. A regular image
2. A more challenging image


## Step 0 – Import necessary libraries

```
# Import necessary libraries
import dlib                      # Face detection library
import cv2                       # Image processing
import matplotlib.pyplot as plt  # Visualization
import numpy as np               # Numerical operations
```

## Step 1 – Load an image (regular image)
Replace 'group_photo.jpg' with your own image path.

```
# Load the image
image_path = 'group_photo.jpg'  # Change to your image path
image = cv2.imread(image_path)

# Convert from BGR (OpenCV) to RGB (Matplotlib)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Display the image
plt.figure(figsize=(8,6))
plt.imshow(image_rgb)
plt.axis('off')
plt.title('Original Image')
plt.show()
```

## Step 2 – Load dlib face detector
We use dlib’s default HOG-based frontal face detector.

```
# Initialize the HOG-based face detector
detector = dlib.get_frontal_face_detector()
```

## Step 3 – Run the detector on the image

```
# Detect faces (1 = upsample once to detect smaller faces)
faces = detector(image_rgb, 1)

print('Number of faces detected:', len(faces))
```

## Step 4 – Draw bounding boxes around detected faces
Each face will have a different random color.

```
# Create a copy of the image to draw on
image_copy = image_rgb.copy()

# Generate random colors for each detected face
colors = [tuple(np.random.randint(0, 255, 3).tolist()) for _ in range(len(faces))]

# Draw rectangles
for i, face in enumerate(faces):
    x1 = face.left()
    y1 = face.top()
    x2 = face.right()
    y2 = face.bottom()

    cv2.rectangle(image_copy, (x1, y1), (x2, y2), colors[i], 3)
    cv2.putText(image_copy, f'Face {i+1}', (x1, y1 - 10),
                cv2.FONT_HERSHEY_SIMPLEX, 0.7, colors[i], 2)

# Show result
plt.figure(figsize=(8,6))
plt.imshow(image_copy)
plt.axis('off')
plt.title('Detected Faces')
plt.show()
```

## Step 5 – Test on a more challenging image
Use an image with:
- Many people
- Smaller faces
- Glasses, hats, masks, helmets
- Different angles


```
# Load challenging image
challenging_path = 'crowd.jpg'  # Change to your challenging image path
challenging_img = cv2.imread(challenging_path)
challenging_rgb = cv2.cvtColor(challenging_img, cv2.COLOR_BGR2RGB)

# Detect faces
faces_challenging = detector(challenging_rgb, 1)
print('Number of faces detected (challenging image):', len(faces_challenging))

# Draw bounding boxes
challenging_copy = challenging_rgb.copy()
colors_ch = [tuple(np.random.randint(0, 255, 3).tolist()) for _ in range(len(faces_challenging))]

for i, face in enumerate(faces_challenging):
    x1 = face.left()
    y1 = face.top()
    x2 = face.right()
    y2 = face.bottom()

    cv2.rectangle(challenging_copy, (x1, y1), (x2, y2), colors_ch[i], 3)
    cv2.putText(challenging_copy, f'Face {i+1}', (x1, y1 - 10),
                cv2.FONT_HERSHEY_SIMPLEX, 0.7, colors_ch[i], 2)

plt.figure(figsize=(10,8))
plt.imshow(challenging_copy)
plt.axis('off')
plt.title('Challenging Image - Detected Faces')
plt.show()
```

## Conclusion
The HOG-based dlib detector works well for frontal, well-lit faces.
However, performance decreases when:
- Faces are very small
- Faces are rotated
- Faces are partially occluded
- Lighting conditions are poor
